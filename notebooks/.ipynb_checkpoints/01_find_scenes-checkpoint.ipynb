```python
from pystac_client import Client
```

```python
catalogue = Client.open('https://earth-search.aws.element84.com/v1')
```

```python
print(catalogue.title)
```

```python
bbox=[88.13, 27.86, 88.23, 27.96]
```

```python
search = catalogue.search(
    collections=["sentinel-2-l2a"],
    bbox=bbox,
    datetime="2022-09-01/2022-11-30",
    query={"eo:cloud_cover": {"lt": 20}}
)
```

```python
item = next(catalogue.get_collection("sentinel-2-l2a").get_all_items())
```

```python
print(item)
```

```python
items = list(search.items())
```

```python
print(len(items))
```

```python
for item in items:
    print(item.id,item.datetime,item.properties['eo:cloud_cover'])
```
